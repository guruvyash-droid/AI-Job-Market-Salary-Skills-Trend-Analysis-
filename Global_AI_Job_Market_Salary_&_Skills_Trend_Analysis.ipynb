```python
skills = (
    df["required_skills"]
    .str.split(",")
    .explode()
    .str.strip()
)

skill_demand = skills.value_counts()

skill_demand.head(15)
```

```python
# skill vs salary
skill_salary = []

for skill in skill_demand.index:
    skill_jobs = df[
        df["required_skills"]
        .str.split(",")
        .apply(lambda x: skill in [s.strip() for s in x])
    ]

    skill_salary.append({
        "skill": skill,
        "job_count": len(skill_jobs),
        "average_salary": skill_jobs["salary_usd"].mean()
    })

skill_salary_df = pd.DataFrame(skill_salary)

skill_salary_df = (
    skill_salary_df
    .sort_values("average_salary", ascending=False)
    .round(2)
)

skill_salary_df.head(15)
```

```python
top_skill_salary = skill_salary_df.head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_skill_salary,
    x="average_salary",
    y="skill"
)

plt.title("Average Salary by Top AI Skills")
plt.xlabel("Average Salary (USD)")
plt.ylabel("Skill")
plt.tight_layout()
plt.show()
```
